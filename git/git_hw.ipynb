```python


import random
def gen_random_list(a: int, b: int, length: int):
   """Функция прнимает три параметра (a: int, b: int, length: int) и возвращает список (list) длины length из равномерно распределенной случайной величины между числами a и b"""
   return [random.randint(a, b) for _ in range(length)]

a_list = gen_random_list(2,34,40)
print(list)
```

```python

def build_frequency_dict(data_list: list):
    """Функция принимает созданный список (list) и считать количество уникальных значений переменной (строить частотный ряд) в виде dict. Например, для списка [1, 2, 3, 1, 2, 2] будет возвращать {1 : 2, 2 : 3, 3 : 1}"""
    frequency = {}
    for i in data_list:
        if i in frequency:
            frequency[i] += 1
        else:
            frequency[i]= 1
    return frequency

frequency_list = build_frequency_dict(a_list)

print(frequency_list)


```

```python

def minimum(lst):
    """Возвращает минимальное значение списка"""
    if not lst:
        return None
    min_val = lst[0]
    for x in lst[1:]:
        if x < min_val:
            min_val = x
    return min_val

def maximum(lst):
    """Возвращает максимальное значение списка."""
    if not lst:
        return None
    max_val = lst[0]
    for x in lst[1:]:
        if x > max_val:
            max_val = x
    return max_val

def mean(lst):
    """Возвращает среднее арифметическое списка."""
    if not lst:
        return None
    return sum(lst) / len(lst)

def median(lst):
    """Возвращает медиану списка (средний элемент или среднее двух средних)."""
    if not lst:
        return None
    sorted_lst = sorted(lst)
    n = len(sorted_lst)
    mid = n // 2
    if n % 2 == 1:
        return sorted_lst[mid]
    else:
        return (sorted_lst[mid - 1] + sorted_lst[mid]) / 2

def mode(lst):
    """
    Возвращает список максимальных значений.
    Если все значения уникальны, возвращает пустой список.
    """
    if not lst:
        return []
    # Подсчёт частот
    freq = {}
    for x in lst:
        freq[x] = freq.get(x, 0) + 1
    max_freq = max(freq.values())
    # Если максимальная частота == 1, то моды нет (все уникальны)
    if max_freq == 1:
        return []
    # Собираем все элементы с максимальной частотой
    modes = [x for x, cnt in freq.items() if cnt == max_freq]
    return modes

print("Минимум:", minimum(a_list))
print("Максимум:", maximum(a_list))
print("Среднее:", mean(a_list))
print("Медиана:", median(a_list))
print("Мода:", mode(a_list))
```
